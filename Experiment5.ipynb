```python

# Medical Symptom Checker
# Knowledge Representation + Rule-Based Reasoning

# 1. Knowledge Base: IF-THEN rules
knowledge_base = [
    {
        "id": "R1",
        "symptoms": {"sneezing", "runny nose"},
        "condition": "Common cold-like illness",
        "explanation": "Sneezing and a runny nose were reported."
    },
    {
        "id": "R2",
        "symptoms": {"fever", "body aches", "fatigue"},
        "condition": "Influenza-like illness",
        "explanation": "Fever, body aches, and fatigue were reported."
    },
    {
        "id": "R3",
        "symptoms": {"fever", "cough", "sore throat"},
        "condition": "Respiratory infection pattern",
        "explanation": "Fever, cough, and sore throat were reported."
    },
    {
        "id": "R4",
        "symptoms": {"itchy eyes", "sneezing", "runny nose"},
        "condition": "Allergy-like pattern",
        "explanation": "Itchy eyes, sneezing, and a runny nose were reported."
    },
    {
        "id": "R5",
        "symptoms": {"vomiting", "diarrhea", "stomach cramps"},
        "condition": "Gastrointestinal illness pattern",
        "explanation": "Vomiting, diarrhea, and stomach cramps were reported."
    },
    {
        "id": "R6",
        "symptoms": {"frequent urination", "increased thirst"},
        "condition": "Symptoms requiring assessment for blood sugar problems",
        "explanation": "Frequent urination and increased thirst were reported."
    }
]

# 2. Safety checks for urgent symptoms
urgent_symptoms = {
    "difficulty breathing",
    "chest pain",
    "confusion",
    "blue lips",
    "severe bleeding"
}

# 3. Inference Engine
def reason_about_symptoms(patient_symptoms):
    results = []

    for rule in knowledge_base:
        # Forward chaining: check facts against each rule
        if rule["symptoms"].issubset(patient_symptoms):
            results.append(rule)

    return results


# 4. Main Agent
def medical_agent():
    print("=" * 48)
    print("      AI MEDICAL SYMPTOM CHECKER")
    print("=" * 48)

    print("\nEnter symptoms separated by commas.")
    print("Example: fever, cough, sore throat")
    print("\nAvailable examples:")
    print(
        "fever, cough, sore throat, body aches, fatigue,"
        "\nsneezing, runny nose, itchy eyes, vomiting,"
        "\ndiarrhea, stomach cramps, frequent urination,"
        "\nincreased thirst, difficulty breathing, chest pain,"
        "\nconfusion, blue lips, severe bleeding"
    )

    user_input = input("\nEnter your symptoms: ").lower()

    # Knowledge representation: symptoms become facts
    patient_symptoms = {
        symptom.strip()
        for symptom in user_input.split(",")
        if symptom.strip()
    }

    if not patient_symptoms:
        print("\nNo symptoms entered. Please try again.")
        return

    print("\n--- Reasoning Results ---")

    # 5. Safety check before displaying possible conditions
    if patient_symptoms & urgent_symptoms:
        print("URGENT: A potentially serious symptom was reported.")
        print("Seek emergency medical care now.")
        print("Do not rely on this symptom checker.")
        return

    # 6. Apply the rule-based inference engine
    matched_rules = reason_about_symptoms(patient_symptoms)

    if matched_rules:
        print("\nPossible patterns to consider:")

        for rule in matched_rules:
            print(f"\nRule matched: {rule['id']}")
            print(f"Possible condition: {rule['condition']}")
            print(f"Reason: {rule['explanation']}")
            print(
                "Matched symptoms:",
                ", ".join(sorted(rule["symptoms"]))
            )
    else:
        print("No predefined rule matched all entered symptoms.")
        print("This does not mean that you are healthy.")

    # 7. General safety message
    print("\nIMPORTANT:")
    print("These results are educational, not a diagnosis.")
    print("Symptoms can have many different causes.")
    print("Consult a qualified healthcare professional.")
    print("Seek care promptly if symptoms are severe or worsening.")


# Run the intelligent agent
if __name__ == "__main__":
    medical_agent()
```

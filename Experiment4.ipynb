```python

import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output
from google import genai

# =====================================================
# 1. API CONFIGURATION
# =====================================================

# Paste your NEW Gemini API key between the quotes.
GEMINI_API_KEY = ""

client = genai.Client(api_key=GEMINI_API_KEY)
MODEL_NAME = "gemini-3.8-flash"

# =====================================================
# 2. COLLEGE INFORMATION
# Replace sample details with official college details.
# =====================================================

COLLEGE_INFO = """
College Name: ABC Engineering College

COURSES OFFERED:
1. B.Tech Computer Science and Engineering (CSE)
2. B.Tech Electronics and Communication Engineering (ECE)
3. B.Tech Electrical and Electronics Engineering (EEE)
4. Master of Business Administration (MBA)

SAMPLE ANNUAL TUITION FEES:
B.Tech CSE: Rs. 1,00,000
B.Tech ECE: Rs. 90,000
B.Tech EEE: Rs. 85,000
MBA: Rs. 80,000

ELIGIBILITY:
B.Tech: 10+2 with Physics, Mathematics, and other
subjects required under the applicable admission rules.
MBA: Bachelor's degree from a recognized university.
Entrance examinations and additional requirements may apply.

ADMISSION PROCEDURE:
1. Check course eligibility.
2. Complete the appropriate admission application.
3. Submit the required documents.
4. Attend counselling or selection, if applicable.
5. Complete document verification.
6. Pay the applicable fees.

DOCUMENTS THAT MAY BE REQUIRED:
- 10th class marks memo
- 12th class marks memo
- Entrance examination rank card, if applicable
- Transfer certificate
- Identity proof
- Passport-size photographs
- Category certificate, if applicable
- Other documents specified by the college

IMPORTANT:
These are fictional sample details for demonstration.
They are not verified fees or rules of a real college.
"""

# =====================================================
# 3. PROMPT ENGINEERING
# =====================================================

SYSTEM_PROMPT = f"""
You are a friendly and helpful College Admission Chatbot.

Your responsibilities:
- Answer questions about courses.
- Explain eligibility criteria.
- Provide available fee information.
- Explain admission procedures.
- List required documents.
- Help students understand the admission process.

COLLEGE INFORMATION:
{COLLEGE_INFO}

RULES:
1. Use simple, clear English.
2. Use the provided college information for college facts.
3. Never invent fees, cutoffs, deadlines, or admission rules.
4. If information is missing, say it is not available.
5. Suggest contacting the official admission office when needed.
6. Explain procedures step by step.
7. Use bullet points when appropriate.
8. Remember previous messages for follow-up questions.
9. Never claim that admission or an application is confirmed.
10. Remind students to verify information officially.
11. Politely redirect unrelated questions to admissions.
"""

# =====================================================
# 4. CONVERSATION HISTORY
# =====================================================

messages = []

# =====================================================
# 5. GENERATE CHATBOT RESPONSE
# =====================================================

def get_response(question):

    messages.append({
        "role": "user",
        "content": question
    })

    conversation = "\n".join(
        ("Student: " if msg["role"] == "user" else "Assistant: ")
        + msg["content"]
        for msg in messages
    )

    full_prompt = f"""
{SYSTEM_PROMPT}

CONVERSATION HISTORY:
{conversation}

Answer the student's latest question.
"""

    try:
        response = client.models.generate_content(
            model=MODEL_NAME,
            contents=full_prompt
        )

        answer = response.text

        if not answer:
            answer = "No response was generated. Please try again."

        messages.append({
            "role": "assistant",
            "content": answer
        })

        return answer

    except Exception as error:
        messages.pop()

        print("Technical details:", str(error))

        return (
            "Sorry, I could not generate a response. "
            "Please check your API key, model availability, "
            "internet connection, and API quota."
        )

# =====================================================
# 6. CHATBOT INTERFACE
# =====================================================

title = widgets.HTML(
    value="""
    <div style="text-align:center; padding:10px;">
        <h1>🎓 College Admission Chatbot</h1>
        <p>Ask about courses, eligibility, fees,
        documents, and admission procedures.</p>
    </div>
    """
)

question_box = widgets.Text(
    placeholder="Type your question here...",
    description="You:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="80%")
)

send_button = widgets.Button(
    description="Send",
    button_style="primary",
    icon="paper-plane",
    layout=widgets.Layout(width="90px")
)

clear_button = widgets.Button(
    description="Clear Chat",
    button_style="warning",
    icon="trash"
)

chat_output = widgets.Output(
    layout=widgets.Layout(
        border="1px solid #cccccc",
        padding="12px",
        width="100%"
    )
)

# =====================================================
# 7. SEND MESSAGE
# =====================================================

def send_message(_=None):

    question = question_box.value.strip()

    if not question:
        return

    question_box.value = ""
    send_button.disabled = True

    try:
        with chat_output:
            display(Markdown(f"**🧑 Student:** {question}"))

            answer = get_response(question)

            display(Markdown(f"**🤖 Admission Bot:**\n\n{answer}"))
            display(Markdown("---"))

    finally:
        send_button.disabled = False

# =====================================================
# 8. CLEAR CONVERSATION
# =====================================================

def clear_chat(_=None):

    messages.clear()

    with chat_output:
        clear_output(wait=True)

# =====================================================
# 9. CONNECT BUTTONS
# =====================================================

send_button.on_click(send_message)
question_box.on_submit(send_message)
clear_button.on_click(clear_chat)

# =====================================================
# 10. DISPLAY CHATBOT
# =====================================================

display(title)

display(
    widgets.HBox(
        [question_box, send_button],
        layout=widgets.Layout(width="100%")
    )
)

display(clear_button)
display(chat_output)

display(Markdown(
    "**Note:** Sample college details are fictional. "
    "Verify fees and admission rules using official sources."
))
```

```python
import os
from getpass import getpass

os.environ["GEMINI_API_KEY"] = getpass("Enter Gemini API Key: ")
os.environ["SERPER_API_KEY"] = getpass("Enter Serper API Key: ")
```

```python
from crewai import LLM

llm = LLM(
    model="gemini/gemini-3.5-flash",
    api_key=os.environ["GEMINI_API_KEY"],
    temperature=0.5
)

print("Gemini initialized successfully!")
```

```python
response = llm.call(
    messages=[
        {
            "role": "user",
            "content": "Explain Artificial Intelligence in Healthcare in 5 lines."
        }
    ]
)

print(response)
```

```python
from crewai_tools import SerperDevTool

search_tool = SerperDevTool()
print("Serper search tool initialized!")
```

```python
from crewai import Agent

research_agent = Agent(
    role="Research Specialist",
    goal="Collect accurate and relevant information about the given research topic.",
    backstory="You are an expert researcher who gathers information from reliable online sources.",
    tools=[search_tool],
    llm=llm,
    verbose=True
)

writer_agent = Agent(
    role="Academic Research Paper Writer",
    goal="Write a well-structured academic research paper using the collected information.",
    backstory="You are an academic writer experienced in preparing technical research papers.",
    llm=llm,
    verbose=True
)

editor_agent = Agent(
    role="Research Paper Editor",
    goal="Review and improve the research paper for quality, grammar, and clarity.",
    backstory="You are an experienced academic editor who ensures papers are professional and consistent.",
    llm=llm,
    verbose=True
)

print("All three agents created successfully!")
```

```python

import asyncio
from crewai import Crew, Process

research_crew = Crew(
    agents=[
        research_agent,
        writer_agent,
        editor_agent
    ],
    tasks=[
        research_task,
        writing_task,
        editing_task
    ],
    process=Process.sequential,
    verbose=True
)

topic = "Artificial Intelligence in Healthcare"

# Retry if Google's server is temporarily unavailable
async def run_crew_with_retry():
    for attempt in range(1, 6):
        try:
            print(f"\nCrew execution attempt {attempt}/5")

            result = await research_crew.kickoff_async(
                inputs={"topic": topic}
            )

            print("\nResearch paper generated successfully!")
            return result

        except Exception as e:
            error = str(e)

            if "503" in error or "UNAVAILABLE" in error:
                if attempt < 5:
                    delay = 2 ** attempt
                    print(f"Gemini is temporarily busy.")
                    print(f"Retrying in {delay} seconds...")
                    await asyncio.sleep(delay)
                else:
                    print("Gemini is still unavailable.")
                    raise
            else:
                raise

result = await run_crew_with_retry()

print(result)
```

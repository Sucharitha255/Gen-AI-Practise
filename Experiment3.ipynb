```python

class SmartHomeAgent:
    def __init__(self):
        self.ac_on = False
        self.target_temp = 24
        self.upper_limit = 26
        self.lower_limit = 24

    def perceive(self, temperature):
        """Receive the room temperature from the sensor."""
        return temperature

    def act(self, temperature):
        """Choose an action based on the temperature."""
        temperature = self.perceive(temperature)

        if temperature > self.upper_limit:
            self.ac_on = True
        elif temperature < self.lower_limit:
            self.ac_on = False

        if self.ac_on:
            print(f"Room: {temperature}°C | AC: ON")
        else:
            print(f"Room: {temperature}°C | AC: OFF")


# Create the intelligent agent
agent = SmartHomeAgent()

# Simulated sensor readings
temperatures = [22, 25, 27, 28, 25, 23, 22, 26, 27]

for temp in temperatures:
    agent.act(temp)
```
